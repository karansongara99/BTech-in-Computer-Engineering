### 4. 📏 Scaling Numerical Data (StandardScaler)
**Concept:** `Installs` (Millions) are much larger than `Rating` (1-5). We scale them so the model treats them equally.

**Task:** Use `StandardScaler` on the numerical columns.

```python
#StandardScaler()
from sklearn.preprocessing import StandardScaler
nums = ['Reviews','Size','Installs','Price']
scaler = StandardScaler()
scalED_numeric = scaler.fit_transform(df[nums])
scalED_numeric[:5]
#fit_transform()
```

### 5. 🔠 Encoding Categorical Data

```python
#get_dummies 
# Using pandas get_dummies to encode categorical columns
df_dummies = pd.get_dummies(df, columns=['Category', 'Content Rating'])
df_dummies.head()
```

**Method B: Sklearn `OneHotEncoder` (Professional)**

```python
#OneHotEncoder
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')
cat_data = df[['Category', 'Content Rating']]
encoded = encoder.fit_transform(cat_data)
# To see the encoded array (first 5 rows)
encoded[:5]
# To get the feature names:
encoder.get_feature_names_out(['Category', 'Content Rating'])
```
