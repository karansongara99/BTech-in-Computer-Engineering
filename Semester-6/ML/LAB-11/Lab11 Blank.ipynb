# Step 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

# Step 6: Distribution of numerical features

```python
plt.figure(figsize=(15, 5))

print("Math Score")
plt.subplot(1, 3, 1)
sns.histplot(df["math score"], kde=True, color="red")
plt.title("Math Score")

print("Reading Score")
plt.subplot(1, 3, 2)
sns.histplot(df["reading score"], kde=True, color="green")
plt.title("Reading Score")

print("Writing Score")
plt.subplot(1, 3, 3)
sns.histplot(df["writing score"], kde=True,color="blue")
plt.title("Writing Score")
plt.show()
```

# Step 7: Apply StandardScaler

```python
X = df[["math score", "reading score", "writing score"]]
scalar = StandardScaler()
X_scaled = scalar.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
print(X_scaled_df.head())
```

# Step 8: Elbow method to find optimal k

```python
wcss = []
k_ranges = range(1, 11)
for k in k_ranges:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled_df)
    wcss.append(kmeans.inertia_)
plt.plot(k_ranges, wcss,marker='o', color='red', linewidth=3)
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```
