# Import diabetes.csv dataset 

```python
df = pd.read_csv("diabetes.csv")
df
```

# Check the distribution of the target

```python
print(df['Outcome'].value_counts())
```

# Replace zeros with NaN for specific columns

```python
cols = ['BloodPressure', 'SkinThickness', 'Insulin','Pregnancies']
df[cols] = df[cols].replace(0, np.nan)
df
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
for col in cols:
    df[col] = df[col].fillna(df[col].median())

# Verify no missing values remain
df.isnull().sum()
```

```python
df.isnull().sum()
```

# Visualize Distributions

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

# Convert Targer data into interger code

```python
# Ensure Outcome is numeric (0/1). If it's object, map text labels, otherwise leave as is.
if df['Outcome'].dtype == 'object':
    df['Outcome'] = df['Outcome'].map({'Diabetic':1,'Non Diabetic':0})

# Convert to integer type (safe if already numeric)
df['Outcome'] = df['Outcome'].astype(int)

# Visualize correlations
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
df['Outcome'] = df['Outcome'].astype(int)
df.dtypes
df
```

# Divide the data into input and output

```python
# Divide the data into input (X) and output (y)
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# Quick peek
X.head()
```

```python
print("Shape:", X.shape)
print("Target distribution:")
print(y.value_counts())
```

# Splitting the dataset into the Training set and Test set

```python
# Splitting the dataset into the Training set and Test set (stratify to preserve target distribution)
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
print("Train shape:", x_train.shape)
print("Test shape:", x_test.shape)
```

# Create Model

```python
# Create the Decision Tree model
model = DecisionTreeClassifier(random_state=42)
```

# Fitting DecisionTreeClassifier   on dataset

```python
# Fit the model to the training data
model.fit(x_train, y_train)
```

# Display Decision Tree

```python
# Display the trained decision tree
from sklearn import tree
plt.figure(figsize=(14,10))
tree.plot_tree(model, feature_names=X.columns, class_names=['Non Diabetic','Diabetic'],
               filled=True, rounded=True, fontsize=10)
plt.show()
```

# Predict the x_test 

```python
# Predict on the test set
y_pred = model.predict(x_test)
# show first few predictions
y_pred[:10]
```

# Display Training Accuracy

```python
# Display Training Accuracy
print("Training accuracy:", model.score(x_train, y_train))
```

# Display Test Accuracy

```python
# Display Test Accuracy
print("Test accuracy:", model.score(x_test, y_test))
```

# Confusion Matrix

```python
# Confusion Matrix and Classification Report
from sklearn.metrics import confusion_matrix, classification_report
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

print(classification_report(y_test, y_pred))
```

```python
# Feature importance
fi = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
fi.plot(kind='bar')
plt.title('Feature Importances')
plt.show()
print(fi)
```
