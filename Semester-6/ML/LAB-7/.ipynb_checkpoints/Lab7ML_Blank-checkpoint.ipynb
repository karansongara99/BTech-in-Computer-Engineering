# Decision Tree Classifier

```python
from sklearn.datasets import load_iris 
```

# Import iris data ser using sklearn

```python
data = load_iris()
```

```python
data
```

# Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
```

# Divide the data into input and output

```python
x=data['data']
x
```

```python
y=data['target']
y
```

# Splitting the dataset into the Training set and Test set

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.3,random_state=1)
```

```python
plt.plot(x,y)
plt.title("This Graph for X and Y")
plt.show()
```

```python
plt.pie(y)
plt.title("Pie Chart For Output")
plt.show()
```

# Fitting DecisionTreeClassifier   on dataset

```python
model=DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

# Display Decision Tree

```python
from sklearn import tree
```

```python
tree.plot_tree(model)
plt.show()
```

# Predict the x_test 

```python
x_pred=model.predict(x_test)
x_pred
```

```python
y_test
```

```python
model.score(x_train,y_train)
```

```python
y_test
```

```python
model.score(x_test,y_test)
```

# Import diabetes.csv dataset 

```python
df = pd.read_csv("diabetes.csv")
df
```

# Check the distribution of the target

```python
print(df['Outcome'].value_counts())
```

# Replace zeros with NaN for specific columns

```python
cols = ['BloodPressure', 'SkinThickness', 'Insulin','Pregnancies']
df[cols] = df[cols].replace(0, np.nan)
df
```

# Check for missing values

```python
df.isnull().sum()
```

# Fill missing values with median

```python
for col in cols:
    df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

# Visualize Distributions

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

# Convert Targer data into interger code

```python
df['Outcome'] = df['Outcome'].map({'Diabetic':1,'Non Diabetic':0})

sns.heatmap(df.corr(),annot=True,cmap='coolwarm')
plt.show()
```

```python
df['Outcome'] = df['Outcome'].astype(int)
df.dtypes
df
```
