# Importing the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

# Importing the libraries

```python
df = pd.read_csv("Position_Salaries.csv")
```

```python
df
```

# Plot the data using scatter plot

```python
X = df[['Level']]
X
```

```python
y = df[['Salary']]
y
```

```python
plt.scatter(X, y, color='blue')
plt.title('Linear Regression')
plt.xlabel('Level')
plt.ylabel('Salary')
plt.show()
```

# Splitting the dataset into the Training set and Test set

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,test_size=0.3)
```

```python
# print(X_train)
print((X_test))
# print(y_test)
# print(y_train)
```

# Fitting Linear Regression to the dataset

```python
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

# Predict the x_test using Linear Model

```python
pr = model.predict(X_test)
pr
```

# Visualising the Linear Regression results

```python
plt.scatter(X_test, y_test)
plt.plot(X_test, pr)
```

# Create Grid for higher resolution and smoother curve

```python

poly = PolynomialFeatures(degree=4, include_bias=False)  # change degree if needed
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

# Smooth curve for plotting
X_grid = np.linspace(X.min(), X.max(), 200).reshape(-1,1)
y_pred = model.predict(poly.transform(X_grid))

plt.scatter(X, y, s=30, label='data')
plt.plot(X_grid, y_pred, color='red', label='polynomial fit')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```
